```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
num_cols = df.select_dtypes(include=np.number).columns
```

```python
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
cat_cols = df.select_dtypes(include="object").columns
```

```python
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"])
```

```python
for col in df.columns:
    print(col)
    print(df[col].unique()[:10])
    print("----------------")
```

```python
Q1 = df["Expense_Amount"].quantile(0.25)
Q3 = df["Expense_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

df = df[(df["Expense_Amount"]>=lower) &
        (df["Expense_Amount"]<=upper)]
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Expense_Amount"], bins=30)
plt.title("Expense Amount Distribution")
plt.xlabel("Expense")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Expense_Category"].value_counts().plot(kind="bar")

plt.title("Expense Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

```python
monthly = df.groupby("Month")["Expense_Amount"].sum()

monthly.plot(kind="line", marker="o")

plt.title("Monthly Expense")

plt.show()
```
